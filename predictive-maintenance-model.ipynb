```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("Libraries loaded successfully.")
```

```python
np.random.seed(42)

n_samples = 1000

previous_cpu = np.random.uniform(5, 85, n_samples)
cpu_change = np.random.normal(5, 10, n_samples)

current_cpu = np.clip(previous_cpu + cpu_change, 1, 100)

cpu_average = np.clip(
    (previous_cpu + current_cpu) / 2 + np.random.normal(0, 3, n_samples),
    1,
    100
)

cpu_trend = current_cpu - previous_cpu

# Simulate the next CPU interval
next_cpu = np.clip(
    current_cpu + (cpu_trend * 0.7) + np.random.normal(0, 6, n_samples),
    1,
    100
)

# Target:
# 1 = CPU exceeds 80% during next interval
# 0 = CPU remains below 80%
high_cpu_next = (next_cpu >= 80).astype(int)

data = pd.DataFrame({
    "previous_cpu": previous_cpu,
    "current_cpu": current_cpu,
    "cpu_average": cpu_average,
    "cpu_trend": cpu_trend,
    "high_cpu_next": high_cpu_next
})

data.head(10)
```

```python
data["high_cpu_next"].value_counts()
```

```python
X = data[
    [
        "previous_cpu",
        "current_cpu",
        "cpu_average",
        "cpu_trend"
    ]
]

y = data["high_cpu_next"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

print("Model training complete.")
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
predictions = model.predict(X_test)

accuracy = accuracy_score(y_test, predictions)
precision = precision_score(y_test, predictions)
recall = recall_score(y_test, predictions)
f1 = f1_score(y_test, predictions)

print(f"Accuracy:  {accuracy:.2%}")
print(f"Precision: {precision:.2%}")
print(f"Recall:    {recall:.2%}")
print(f"F1 Score:  {f1:.2%}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, predictions))

print("\nClassification Report:")
print(classification_report(y_test, predictions))
```

```python
test_scenario = pd.DataFrame({
    "previous_cpu": [55],
    "current_cpu": [70],
    "cpu_average": [62.5],
    "cpu_trend": [15]
})

prediction = model.predict(test_scenario)[0]
probability = model.predict_proba(test_scenario)[0][1]

print("Current CPU: 70%")
print("Previous CPU: 55%")
print("CPU Trend: +15%")

print("\n--- Predictive Maintenance Result ---")

if prediction == 1:
    print("Status: HIGH CPU RISK")
    print("Recommended Action: Preventive maintenance alert")
else:
    print("Status: NORMAL")
    print("Recommended Action: Continue monitoring")

print(f"High CPU Probability: {probability:.2%}")
```
